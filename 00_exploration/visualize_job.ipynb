```python
import pandas as pd
import pyarrow.parquet as pq
import pyarrow.compute as pc
```

```python
dataset = pq.ParquetDataset(
    '../dataset/prom_slurm_joined/prom_slurm_joined.parquet',
    filters=[('gpu_node', '=', 1)]
)

fragments = []
for fragment in dataset.fragments:
    batch = fragment.to_table(
        filter=pc.field('gpu_node') == 1
    ).to_pandas()
    fragments.append(batch)
 
gpu_df = pd.concat(fragments).reset_index(drop=True)
print(f"GPU nodes: {len(gpu_df):,} linhas")
print(f"Nodes disponíveis: {gpu_df['node'].unique()}")
print(f"Jobs disponíveis (slurm_id): {gpu_df['slurm_id'].nunique()} únicos")
```

```python
# ── 2. Escolher um job específico ─────────────────────────────────────────────
JOB_ID = gpu_df["slurm_id"].iloc[0]   # troque pelo slurm_id que quiser
print(f"\nJob escolhido: {JOB_ID}")
```

```python
# ── 3. Pegar todas as linhas desse job ────────────────────────────────────────
job_df = gpu_df[gpu_df["slurm_id"] == JOB_ID].sort_values("timestamp")
 
print(f"Snapshots do job: {len(job_df)}")
print(f"Node: {job_df['node'].iloc[0]}")
print(f"Estado: {job_df['state'].iloc[0]}")
print(f"Início: {job_df['start_date'].iloc[0]}")
print(f"Fim:    {job_df['end_date'].iloc[0]}")
print()
print(job_df.head())
 
```
